Creating 2 pairs of 2 for our pro forma (a document that shows anticipated financial results of x event or transaction) pairs

```python
# import em
import pandas as pd
import numpy as np
import random
```

```python
# assign em
us = ['Nick', 'Jake', 'Sudheesh', 'Tom']

# shuffle
random.shuffle(us)

# divide into two groups
group1 = us[:2]
group2 = us[2:]

# print em
print(f'group one is: {group1}')
print(f'group two is: {group2}')
```
